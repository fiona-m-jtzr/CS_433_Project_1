# Imports

```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd
from sklearn.metrics import confusion_matrix
import seaborn as sns
import statsmodels.formula.api as smf
import sklearn
```

# 1-Opening data

```python
#Train
#Inputs
x_train = np.genfromtxt(
    "x_train.csv",
    delimiter=",",
    skip_header=1
)
print("Train input : OK")

#Outputs
y_train = np.genfromtxt(
    "y_train.csv",
    delimiter=",",
    skip_header=1
)
print("Train output : OK")

#Test
#Inputs
x_test = np.genfromtxt(
    "x_test.csv",
    delimiter=",",
    skip_header=1
)
print("Test input : OK")
```

# 2-Data cleaning

```python
#Keep track of x_train size
N = len(x_train)

#Train/test input stack
x_train_test = np.vstack([x_train, x_test])

#Data cleaning
x_train_test_clean = x_train_test[:, ~np.any(np.isnan(x_train_test), axis=0)]

#Train/test reseparation
x_train_clean = x_train_test_clean[:N, :]
x_test_clean = x_train_test_clean[N:, :]
```

# 3-Data saving

```python
#Save in .npy format
np.save("x_train_clean.npy", x_train_clean)
np.save("x_test_clean.npy", x_test_clean)
np.save("y_train.npy", y_train)
```
